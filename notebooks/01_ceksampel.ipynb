```python
import matplotlib.pyplot as plt, random
from PIL import Image
from pathlib import Path

def show(split, label, n=6):
    files = list(Path(f"../data_cls/{split}/{label}").glob("*"))
    sample = random.sample(files, min(n, len(files)))
    fig, axes = plt.subplots(1, n, figsize=(3*n, 3.2))
    for ax, f in zip(axes, sample):
        ax.imshow(Image.open(f)); ax.set_title(f.name[:22], fontsize=6); ax.axis("off")
    plt.suptitle(f"{split}/{label}"); plt.show()

show("train", "healthy")
show("train", "bleached")
show("train", "dead")   # ⬅️ PALING PENTING: lihat pola kelas dead
```

```python
import matplotlib.pyplot as plt, random
from PIL import Image
from pathlib import Path

def show_per_source(split, label, prefix, n=8):
    files = [f for f in Path(f"../data_cls/{split}/{label}").glob("*")
             if f.name.lower().startswith(prefix)]
    print(f"{prefix}* di {split}/{label}: {len(files)} file")
    if not files:
        return
    random.shuffle(files)
    sample = files[:n]
    fig, axes = plt.subplots(1, len(sample), figsize=(2.6*len(sample), 3.2))
    for ax, f in zip(axes, sample):
        ax.imshow(Image.open(f)); ax.set_title(f.name[:24], fontsize=6); ax.axis("off")
    plt.suptitle(f"{label} ← sumber: {prefix}")
    plt.show()

# Jalankan SATU PER SATU, perhatikan mana yang berisi gambar putih/bleached:
show_per_source("train", "healthy", "dataset1_")
show_per_source("train", "healthy", "dataset2_")
show_per_source("train", "healthy", "dataset3_")
show_per_source("train", "healthy", "dataset6_")
show_per_source("train", "healthy", "dataset7_")
```
